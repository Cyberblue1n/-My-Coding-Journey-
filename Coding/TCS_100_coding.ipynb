## Question 1: Find the smallest number in an array

```python
def small_num(arr):
    min_num = float('inf')
    for i in arr:
        if i < min_num :
            min_num = i
        
    
    return min_num

if __name__ == "__main__":
    arr = [6,8,2,1,5]
    print(small_num(arr))

'''
Time complexity: O(n)
Space Complexity: O(1)
'''
```

## Question 2: Find the largest number in an array

```python
def largest_num(arr):
    max_num = float('-inf')
    for i in arr:
        if i > max_num :
            max_num = i
        
    
    return max_num

if __name__ == "__main__":
    arr = [6,8,2,1,5]
    print(largest_num(arr))

'''
Time complexity: O(n)
Space Complexity: O(1)
'''
```

## Question 3: Second Smallest and Second Largest element in an array

```python
def second_small(arr):
    if len(arr)<2:
        return None
    
    min_num = float('inf')
    min_num_2 = float('inf')
    for i in arr:
        if i<min_num:
            min_num_2 = min_num
            min_num = i
        elif min_num<i and min_num_2>i:
            min_num_2 = i

    return min_num_2

def second_large(arr):
    if len(arr)<2:
        return None
    
    max_num = float('-inf')
    max_num_2 = float('-inf')
    for i in arr:
        if i>max_num:
            max_num_2 = max_num
            max_num = i
        elif max_num>i and max_num_2<i:
            max_num_2 = i
    return max_num_2

if __name__ == "__main__":
    arr = [1,1,2,2,3]
    print("Second small: ",second_small(arr))
    print("Second large: ",second_large(arr))

'''
Time complexity: O(n)
Space Complexity: O(1)
'''
```

## Question 4: Reverse a given array

```python
def reverse_array(arr):
    i=0
    j=len(arr)-1
    while i<j:
        arr[i], arr[j] = arr[j], arr[i]
        i+=1
        j-=1
    
    return arr

if __name__ == "__main__":
    arr = [6,8,2,1,5]
    print(reverse_array(arr))

'''
Time complexity: O(n)
Space Complexity: O(1)
'''
```

## Question 5: Count frequency of each element in an array

```python
def count_frequency(arr):
    count_freq = {}
    for i in arr:
        if i in count_freq:
            count_freq[i]+=1
        
        else:
            count_freq[i] = 1
    
    return count_freq

if __name__ == "__main__":
    arr = [1,2,2,1,5]
    print(count_frequency(arr))

'''
Time complexity: O(n)
Space Complexity: O(n)

Another method:
from collections import Counter
print(Counter([1,2,2,1,5]))
'''
```

## Question 24: Search an element in an array

```python
#Linear search:
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i]==target:
            return i+1

    return i

def binary_search(arr, target):
    i=0
    j=len(arr)-1
    while i<=j:
        mid = (i+j)//2
        if arr[mid] == target:
            return mid+1
        elif arr[mid] > target:
            j = mid+1
        elif arr[mid] < target:
            i = mid-1

    return None

if __name__ =="__main__":
    arr1 = [6,1,4,2,9]
    arr2 = [5,6,7,8,9]
    target = 6
    print("Linear Search",linear_search(arr1, target))
    print("binary Search",binary_search(arr2, target))
'''
Linear search
Time complexity: O(n)
Space Complexity: O(1)

Binary Search
Time complexity: O(log(n))
Space Complexity: O(1)
'''
```

## Question 26: Check if a number is palindrome or not

```python
def check_palindrome(num):
    if reverse_number(num) == num:
        return True
    else:
        return False

def reverse_number(num):
    new_num = 0
    while num>0:
        a = num%10
        new_num = new_num*10 + a
        num = num//10
    return new_num

if __name__ == "__main__":
    number = 121
    print(check_palindrome(number))

'''
Time complexity: O(n)
Space Complexity: O(1)
'''
```

## Question 27: Find all Palindrome numbers in a given range

```python
def palindrome_number(num):
    mynum = num
    b=0
    while num>0:
        a = num%10
        b = b*10 + a
        num = num//10
    
    return b==mynum
    
if __name__ == "__main__":
    n = 25
    for i in range(10, n):
        if palindrome_number(i) == True:
            print(i, end=" ")        
    
'''
Time complexity: O(n) * O(log(n))
Space Complexity: O(1)
'''
```

## Question 40: Reverse digits of a number

```python
def reverse_digits(num):
    b=0
    while num>0:
        b = b*10 + (num%10)
        num = num//10
    
    return b

if __name__ == "__main__":
    num = 12345
    print(reverse_digits(num))

'''
Time complexity: O(log(n))
Space Complexity: O(1)
'''
```

## Question 41: Maximum and Minimum digit in a number

```python
def max_min_num(num):
    num = str(num)
    max_num = float('-inf')
    min_num = float('inf')
    for i in num:
        if max_num < int(i):
            max_num = int(i)
        if min_num > int(i):
            min_num = int(i)

    return max_num, min_num

if __name__ == "__main__":
    num = 12345
    print(max_min_num(num))

'''
Time complexity: O(n)
Space Complexity: O(1)
'''
```

## Question 68: Bubble Sort Algorithm

```python
def bubble_sort(arr):
    for i in range(0, len(arr)-1):
        for j in range(0, len(arr)-i-1):
            if arr[j]>=arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
            
    return arr


if __name__ == "__main__":
    arr = [6,5,3,4,1,2]
    print(bubble_sort(arr))

'''
Time complexity: O(n^2)
Space Complexity: O(1)
'''
```
