# Array

## Question 1: Find the smallest number in an array

```python
def small_num(arr):
    min_num = float('inf')
    for i in arr:
        if i < min_num :
            min_num = i
        
    
    return min_num

if __name__ == "__main__":
    arr = [6,8,2,1,5]
    print(small_num(arr))

'''
Time complexity: O(n)
Space Complexity: O(1)
'''
```

## Question 2: Find the largest number in an array

```python
def largest_num(arr):
    max_num = float('-inf')
    for i in arr:
        if i > max_num :
            max_num = i
        
    
    return max_num

if __name__ == "__main__":
    arr = [6,8,2,1,5]
    print(largest_num(arr))

'''
Time complexity: O(n)
Space Complexity: O(1)
'''
```

## Question 4: Reverse a given array

```python
def reverse_array(arr):
    i=0
    j=len(arr)-1
    while i<j:
        arr[i], arr[j] = arr[j], arr[i]
        i+=1
        j-=1
    
    return arr

if __name__ == "__main__":
    arr = [6,8,2,1,5]
    print(reverse_array(arr))

'''
Time complexity: O(n)
Space Complexity: O(1)
'''
```

## Question 26: Check if a number is palindrome or not

```python
def check_palindrome(num):
    if reverse_number(num) == num:
        return True
    else:
        return False

def reverse_number(num):
    new_num = 0
    while num>0:
        a = num%10
        new_num = new_num*10 + a
        num = num//10
    return new_num

if __name__ == "__main__":
    number = 121
    print(check_palindrome(number))

'''
Time complexity: O(n)
Space Complexity: O(1)
'''
```

## Question 32: Even or Odd

```python
def even_odd(num):
    if num%2==0:
        return "Even"
    else:
        return "Odd"
    
if __name__ == "__main__":
    num = 7
    print(even_odd(num))

'''
Time complexity: O(1)
Space Complexity: O(1)
'''
```

## Question 33: Check weather a given number is positive or negative

```python
def positive_negative(num):
    if num>0:
        return "Positive"
    else:
        return "Negative"
    
if __name__ == "__main__":
    num = -7
    print(positive_negative(num))

'''
Time complexity: O(1)
Space Complexity: O(1)
'''
```

## Question 34: Sum of first N natural numbers

```python
def sum_of_natural(n):
    return (n*(n+1))//2

if __name__ == "__main__":
    n=5
    print(sum_of_natural(n))

'''
Time complexity: O(1)
Space Complexity: O(1)
'''
```
