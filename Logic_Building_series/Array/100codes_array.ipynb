## Counting the number of even and odd elements in an array

```python
def count_even_odd(arr):
    count_odd = 0
    count_even = 0
    for i in arr:
        if i%2==0:
            count_even+=1
        else:
            count_odd+=1
    
    return f"even count: {count_even} and odd count: {count_odd}"

arr = [1, 7, 8, 4, 5, 16, 8]
print(count_even_odd(arr))
```

## Find Largest element in an array

```python
def largest_el(arr):
    large = float('-inf')
    for i in arr:
        if i>large:
            large = i
        
    return large

arr = [1, 7, 8, 4, 5, 16, 8]
print(largest_el(arr))
```

## Find Smallest element in an array

```python
def smallest_el(arr):
    small = arr[0]
    for i in arr:
        if i<small:
            small = i
        
    return small

arr = [1, 7, 8, 4, 5, 16, 8]
print(smallest_el(arr))
```

## Find Second Smallest Element in an Array

```python
def sec_smallest_el(arr):
    small1 = float('inf')
    small2 = float('inf')
    for i in arr:
        if i<small1:
            small2 = small1
            small1 = i
        elif i<small2 and small1<i:
            small2 = i
        
    return small2

arr = [1, 7, 8, 4, 5, 16, 8]
print(sec_smallest_el(arr))
```

## Calculate the sum of elements in an array

```python
def sum_of_el(arr):
    total = 0
    for i in arr:
        total+=i
    
    return total

arr = [1, 7, 8, 4, 5, 16, 8]
print(sum_of_el(arr))
```

## Reverse an array

```python
def rev_arr(arr):
    return arr[::-1]

arr = [1, 7, 8, 4, 5, 16, 8]
print(rev_arr(arr))
```

## Removing Duplicate elements from an array

```python
def remove_dup(arr):
    return set(arr)

arr = [1, 7, 7, 8, 8, 4, 5]
print(remove_dup(arr))
```

## Counting Distinct Elements in an Array

```python
def count_el(arr):
    count = {}
    for i in arr:
        if i in count:
            count[i]+= 1
        else:
            count[i] = 1
        
    return count

arr = [1, 7, 7, 8, 8, 4, 5]
print(count_el(arr))
```

## Finding  Repeating elements in an Array

```python
def repeating_el(arr):
    count = {}
    output = []
    for i in arr:
        if i in count:
            count[i]+=1
        else:
            count[i] = 1

        
    for key, val in count.items():
        if val>1:
            output.append(key)
    
    return output

arr = [1, 7, 7, 8, 8, 4, 5]
print(repeating_el(arr))
```

## Find non-repeating elements in an array 
Example <br>
Input : arr[8] = [10, 20, 70, 90, 80, 20, 10, 20]  <br>
Output : 70 90 80  <br>
Explanation : 70, 90 and 80 are occur 1 time in the given array, 10 occurs 2 times and 20 occurs 3 times.

```python
def non_repeating(arr):
    count_el = {}
    for i in arr:
        if i in arr:
            count_el[i] = 1+count_el.get(i, 0)
        else:
            count_el[i] = 1
        
    for j in count_el.keys():
        if count_el[j]==1:
            print(j)


if __name__ == "__main__":
    arr = [10, 20, 70, 90, 80, 20, 10]
    non_repeating(arr)
        
```

## Finding Minimum scalar product of two vector

```python
def min_scalar_product(arr1, arr2):
    arr1.sort()
    arr2.sort(reverse=True)
    product = 0

    for i in range(len(arr1)):
        product+= arr1[i]*arr2[i]

    return product


if __name__ == "__main__":
    arr1 = [1, 2, 6, 3, 7]
    arr2 = [10, 7, 45, 3, 7]
    print(min_scalar_product(arr1, arr2))
```

## Finding Maximum scalar product of two vectors in an array


```python
def max_scalar_product(arr1, arr2):
    arr1.sort()
    arr2.sort()
    product = 0

    for i in range(len(arr1)):
        product+= arr1[i]*arr2[i]

    return product


if __name__ == "__main__":
    arr1 = [10, 30, 40, 20]
    arr2 = [2, 4, 5, 1]
    print(max_scalar_product(arr1, arr2))
```

## Find all Symmetric pairs in an array

Input : arr = {{10,20}, {30,40}, {50,60}, {20,10}, {40,30}, {90, 100}, {1, 9}, {100, 90}} <br>
Output : (10,20) (30,40) (90, 100)

```python
def symetric_arr(arr):
    output = set()
    for (i,j) in arr:
        output.add((i, j))
    
        if (j, i) in output:
            print((i, j))

if __name__ == "__main__":
    arr = [[10,20], [30,40], [50,60], [20,10], [40,30], [90, 100], [1, 9], [100, 90]]
    symetric_arr(arr)
```

## Finding Arrays are disjoint or not
Enter array1 1 2 3 4 5 <br>
Enter array2 4 5 6 7 8 <br>
Not disjoint


```python
def disjoint_arr(arr1, arr2):
    intersection = []
    for i in arr2:
        if i in arr1:
            intersection.append(i)
        
    return "Disjoint" if intersection == [] else "Not Disjoint"


arr1 = [1,2,3,4,5]
arr2 = [4,5,6,7,8]
print(disjoint_arr(arr1, arr2))
```
