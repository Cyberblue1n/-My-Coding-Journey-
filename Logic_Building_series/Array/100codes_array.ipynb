## Counting the number of even and odd elements in an array

```python
def count_even_odd(arr):
    count_odd = 0
    count_even = 0
    for i in arr:
        if i%2==0:
            count_even+=1
        else:
            count_odd+=1
    
    return f"even count: {count_even} and odd count: {count_odd}"

arr = [1, 7, 8, 4, 5, 16, 8]
print(count_even_odd(arr))
```

## Find Largest element in an array

```python
def largest_el(arr):
    large = float('-inf')
    for i in arr:
        if i>large:
            large = i
        
    return large

arr = [1, 7, 8, 4, 5, 16, 8]
print(largest_el(arr))
```

## Find Smallest element in an array

```python
def smallest_el(arr):
    small = arr[0]
    for i in arr:
        if i<small:
            small = i
        
    return small

arr = [1, 7, 8, 4, 5, 16, 8]
print(smallest_el(arr))
```

## Find Second Smallest Element in an Array

```python
def sec_smallest_el(arr):
    small1 = float('inf')
    small2 = float('inf')
    for i in arr:
        if i<small1:
            small2 = small1
            small1 = i
        elif i<small2 and small1<i:
            small2 = i
        
    return small2

arr = [1, 7, 8, 4, 5, 16, 8]
print(sec_smallest_el(arr))
```

## Calculate the sum of elements in an array

```python
def sum_of_el(arr):
    total = 0
    for i in arr:
        total+=i
    
    return total

arr = [1, 7, 8, 4, 5, 16, 8]
print(sum_of_el(arr))
```

## Reverse an array

```python
def rev_arr(arr):
    return arr[::-1]

arr = [1, 7, 8, 4, 5, 16, 8]
print(rev_arr(arr))
```

## Removing Duplicate elements from an array

```python
def remove_dup(arr):
    return set(arr)

arr = [1, 7, 7, 8, 8, 4, 5]
print(remove_dup(arr))
```

## Counting Distinct Elements in an Array

```python
def count_el(arr):
    count = {}
    for i in arr:
        if i in count:
            count[i]+= 1
        else:
            count[i] = 1
        
    return count

arr = [1, 7, 7, 8, 8, 4, 5]
print(count_el(arr))
```

## Finding  Repeating elements in an Array

```python
def repeating_el(arr):
    count = {}
    output = []
    for i in arr:
        if i in count:
            count[i]+=1
        else:
            count[i] = 1

        
    for key, val in count.items():
        if val>1:
            output.append(key)
    
    return output

arr = [1, 7, 7, 8, 8, 4, 5]
print(repeating_el(arr))
```
