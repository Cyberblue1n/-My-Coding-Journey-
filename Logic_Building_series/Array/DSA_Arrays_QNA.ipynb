```python
#Reverse the array
def reverse_array(arr):
        # code here
    i=0
    j=len(arr)-1
    while i<j:
        arr[i], arr[j] = arr[j], arr[i]
        i+=1
        j-=1
    
    return arr
        
if __name__ == "__main__":
    arr = [1, 4, 3, 2, 6, 5]
    print(reverse_array(arr))
```

```python
#Union of two arrays
def findUnion(a, b):
        # code here 
        output = []
        for i in a:
            if i not in output:
                output.append(i)
        
        for j in b:
            if j not in output:
                output.append(j)
                
        
        output.sort()
        return output


if __name__ == "__main__":
    arr1 = [2,2,3,4,5]
    arr2 = [1,1,2,3,4]
    print(findUnion(arr1, arr2))
```

## Second Largest element in an array

```python
def secondlargest(arr):
    largest = second_largest = float('-inf')
    for num in arr:
        if num>largest:
            second_largest = largest
            largest = num
        elif num>second_largest and largest!=num:
            second_largest = num
        
    return second_largest

if __name__ == "__main__":
    arr = [1,2,3,4,5]
    print(secondlargest(arr))
```

## Find the Duplicate Number

```python
def duplicate_number(arr):
    return len(arr)!=len(set(arr))
if __name__ == "__main__":
    arr = [1,2,3,4,5,4]
    print(duplicate_number(arr))
```

## Array Subset of another array

```python
def subset_array(arr1, arr2):   
    count_arr1 = {}
    count_arr2 = {}
    for i in range(len(arr1)):
        count_arr1[arr1[i]] = 1+count_arr1.get(arr1[i], 0)

    for j in range(len(arr2)):
        count_arr2[arr2[j]] = 1+count_arr2.get(arr2[j], 0)
    
    for num, freq in count_arr1.items():
        if num not in count_arr2 or count_arr2[num] < freq:
            return False
    
    return True

if __name__ == "__main__":
    arr1= [1,2,3,4]
    arr2=[1,2,3,4,5]
    print(subset_array(arr1,arr2))
```
