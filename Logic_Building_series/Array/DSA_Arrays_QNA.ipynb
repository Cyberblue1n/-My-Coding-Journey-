```python
#Union of two arrays
def findUnion(a, b):
        # code here 
        output = []
        for i in a:
            if i not in output:
                output.append(i)
        
        for j in b:
            if j not in output:
                output.append(j)
                
        
        output.sort()
        return output


if __name__ == "__main__":
    arr1 = [2,2,3,4,5]
    arr2 = [1,1,2,3,4]
    print(findUnion(arr1, arr2))
```
