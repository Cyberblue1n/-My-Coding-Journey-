# 1. Counting Distinct Elements in an Array

```python
#using set 
def count_distinct(arr):
    arr = set(arr)
    return len(arr)

if __name__ == "__main__":
    a = [10, 20, 40, 30, 50, 20, 10, 20]
    print(count_distinct(a))
```

```python
#using hashmap

def count_distinct(arr):
    count = {}
    for i in arr:
        if i in count.keys():
            count[i]+=1
        else:
            count[i]=1
    return len(count)

if __name__ == "__main__":
    arr = [10, 20, 40, 30, 50, 20, 10, 20]
    print(count_distinct(arr))
    
```

# 2. Removing Duplicate elements from an array

```python
# set method
def remove_duplicate(arr):
    arr = set(arr)
    return arr

if __name__ == "__main__":
    arr = [10, 20, 20, 30, 40, 40, 40, 50, 50]
    print(remove_duplicate(arr))
```

```python
#list method
def remove_duplicate(arr):
    output = []
    for i in range(len(arr)):
        if arr[i] not in output:
            output.append(arr[i])
        
    return output


if __name__ == "__main__":
    arr = [10, 20, 20, 30, 40, 40, 40, 50, 50]
    print(remove_duplicate(arr))
```
