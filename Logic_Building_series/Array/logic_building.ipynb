## Largest element in an array
Example 1:  
Input: arr[] = {2,5,1,3,0};  
Output: 5 <br>
Explanation: 5 is the largest element in the array.  <br>
<br>
Example2: <br>
Input: arr[] = {8,10,5,7,9}; <br>
Output: 10 <br>
Explanation: 10 is the largest element in the array. <br>

```python
def large_el(arr):
    max_val=float('-inf')
    for i in arr:
        if i>max_val:
            max_val = i
    
    return max_val

arr=[2,5,1,3,0]
print(large_el(arr))
```

## Find Second Smallest and Second Largest Element in an array
Example 1: <br>
Input: [1,2,4,7,7,5] <br>
Output: Second Smallest : 2 <br> Second Largest : 5 <br>
Explanation: The elements are as follows 1,2,3,5,7,7 and hence second largest of these is 5 and second smallest is 2 <br>
<br>
Example 2: <br>
Input: [1] <br>
Output: Second Smallest : -1 <br>
	Second Largest : -1 <br>
Explanation: Since there is only one element in the array, it is the largest and smallest element present in the array. There is no second largest or second smallest element present.<br>

```python
def second_large(arr):
    max_val = float('-inf')
    sec_max_val = float('-inf')
    for i in arr:
        if i>max_val:
            sec_max_val = max_val
            max_val = i

        elif i>sec_max_val and i<max_val:
            sec_max_val = i
        
    if sec_max_val == float('-inf'):
        return -1
    else:
        return sec_max_val

def second_minimum(arr):
    min_val = float('inf')
    sec_min_val = float('inf')

    for j in arr:
        if j<min_val:
            sec_min_val = min_val
            min_val = j
        elif j<sec_min_val and j>min_val:
            sec_min_val = j
        
    if sec_min_val==float('inf'):
        return -1
    else:
        return sec_min_val

arr = [1]
print(second_large(arr))
print(second_minimum(arr))
```

## Check if an Array is Sorted
Example 1: <br>
Input: N = 5, array[] = {1,2,3,4,5} <br>
Output: True. <br>

```python
def check_sort(arr):
    for i in range(len(arr)-1):
        if arr[i] > arr[i+1]:
            return False
    
    return True

arr = [1,2,3,5,5]
print(check_sort(arr))
```
