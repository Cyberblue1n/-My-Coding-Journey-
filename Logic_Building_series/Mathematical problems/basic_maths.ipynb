## Count digits in a number
Example 1: <br>
Input:N = 12345 <br>
Output:5 <br>
Explanation:  The number 12345 has 5 digits. <br>

```python
def count_digit(num):
    count=0
    while num>0:
        num = num//10
        count+=1

    return count

num=12345
print(count_digit(num))
```

## Reverse Digits of A Number
Example 1: <br>
Input:N = 12345 <br>
Output: 54321 <br>
Explanation: The reverse of 12345 is 54321. <br>


```python
def reverse_digit(num):
    b=0
    while num>0:
        rem=num%10
        b=b*10+rem
        num = num//10

    return b

num=12345
print(reverse_digit(num))
```

## Check if a number is Palindrome or Not
Example 1:  <br>
Input:N = 4554  <br>
Output:Palindrome Number <br>
Explanation: The reverse of 4554 is 4554 and therefore it is palindrome number  <br>

```python
def palindrome(num):
    check_num = num
    b=0
    while num>0:
        rem=num%10
        b=b*10+rem
        num = num//10

    return "palindrome number" if check_num==b else "not palindrome"

num=4554
print(palindrome(num))
```

## Find GCD of two numbers
Example 1: <br>
Input:N1 = 9, N2 = 12 <br>


Output:3 <br>
Explanation:Factors of 9: 1, 3 and 9  
Factors of 12: 1, 2, 3, 4, 6, 12  
Common Factors: 1, 3 out of which 3 is the greatest hence it is the GCD.  

```python
def gcd(num1, num2):
    gcd = 1
    for i in range(1, min(num1, num2)+1):
        if num1%i == 0  and num2 % i ==0:
            gcd = i
    
    return gcd

print(gcd(9,12))
```

## Check if a number is Armstrong Number or not
Example 1:  
Input:N = 153  

Output:True  

Explanation: 13+53+33 = 1 + 125 + 27 = 153  

```python
def is_armstrong(num):
    digits = str(num)
    power = len(digits)
    
    total = 0
    for d in digits:
        total += int(d) ** power
    
    return num == total

num = 153
if is_armstrong(num):
    print(f"{num} is an Armstrong number")
else:
    print(f"{num} is not an Armstrong number")
```

## Print all Divisors of a given Number
Example 1:  
Input:N = 36  
Output:[1, 2, 3, 4, 6, 9, 12, 18, 36]  

Explanation: The divisors of 36 are 1, 2, 3, 4, 6, 9, 12, 18, 36.

```python
def divisors(num):
    for i in range(1, num+1):
        if num%i==0:
            print(i,end=" ")

num = 36
divisors(num)
```

## Check if a number is prime or not
Example 1:  
Input:N = 2  
Output:True   
Explanation: 2 is a prime number because it has two divisors: 1 and 2 (the number itself).  

```python
def prime_num(num):
    if num==1 or num==2:
        return True
    
    for i in range(3, int(num**0.5)+1):
        if num%i==0:
            return False
    
    return True

num=5
print(prime_num(num))
```
