## Square Star Pattern
<img src="https://static.takeuforward.org/wp/uploads/2022/08/P1.png" width="100">


```python
l=1
r=5
for i in range(1, 6):
    for j in range(1, 6):
        if j>=l:
            print("*",end=" ")
        else:
            print("")
    print("")

```

## Right-Angled Triangle Pattern
<img src="https://static.takeuforward.org/wp/uploads/2022/08/P2.png" width="100">


```python
l=1
r=1
for i in range(1, 6):
    for j in range(1, 6):
        if j>=l and j<=r:
            print("*",end="")
        else:
            print("",end="")
    
    print("")
    r+=1
```

## Right-Angled Number Pyramid
<img src="https://static.takeuforward.org/wp/uploads/2022/08/P3.png" width="100">

```python
l=1
r=1
for i in range(1, 6):
    count = 1
    for j in range(1, 6):
        if j>=l and j<=r:
            print(f"{count}",end="")
            count+=1
        else:
            print("",end="")
    
    print("")
    r+=1

#<img src="https://static.takeuforward.org/wp/uploads/2022/08/P2.png" width="100">
```

## Inverted Right Pyramid
<img src="http://static.takeuforward.org/wp/uploads/2022/08/P5.png" width="100">

```python
l=1
r=5
for i in range(1, 6):
    for j in range(1, 6):
        if j>=l and j<=r:
            print("*",end="")
        else:
            print("",end="")
    
    print("")
    r-=1
```

## Inverted Numbered Right Pyramid
<img src="https://sillycodes.com/wp-content/uploads/2015/10/Invarted-traingle-number-Pattern-in-c-300x296.png" width="100">

```python
l=1
r=5
for i in range(1, 6):
    num=1
    for j in range(1, 6):
        if j>=l and j<=r:
            print(f"{num}",end="")
            num+=1
        else:
            print("",end="")
    
    print("")
    r-=1
```
