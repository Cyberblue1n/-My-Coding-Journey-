## Square Star Pattern

```python
n=4
for i in range(n):
    for j in range(n):
        print("*",end="")
    print()
```

## Hollow Square Star Pattern

```python
n=4
for i in range(1, n+1):
    for j in range(1, n+1):
        if (i==1 or j==1 or i==n or j==n):
            print("*", end="")
        else:
            print(" ", end="")
    print()
```
