## Square Star Pattern
<img src="https://static.takeuforward.org/wp/uploads/2022/08/P1.png" width="100">


```python
l=1
r=5
for i in range(1, 6):
    for j in range(1, 6):
        if j>=l:
            print("*",end=" ")
        else:
            print("")
    print("")

```

## Right-Angled Triangle Pattern
<img src="https://static.takeuforward.org/wp/uploads/2022/08/P2.png" width="100">


```python
l=1
r=1
for i in range(1, 6):
    for j in range(1, 6):
        if j>=l and j<=r:
            print("*",end="")
        else:
            print("",end="")
    
    print("")
    r+=1
```

## Right-Angled Number Pyramid
<img src="https://static.takeuforward.org/wp/uploads/2022/08/P3.png" width="100">

```python
l=1
r=1
for i in range(1, 6):
    count = 1
    for j in range(1, 6):
        if j>=l and j<=r:
            print(f"{count}",end="")
            count+=1
        else:
            print("",end="")
    
    print("")
    r+=1

#<img src="https://static.takeuforward.org/wp/uploads/2022/08/P2.png" width="100">
```

## Increasing Letter Triangle Pattern
<img src="https://static.takeuforward.org/wp/uploads/2023/02/image-18.png" width="100">

```python
l=1
r=1
for i in range(1, 6):
    num=65
    for j in range(1, 6):
        if j>=l and j<=r:
            print(f"{chr(num)}",end="")
            num+=1
        else:
            print("",end="")
    
    print("")
    r+=1
```

## Alpha-Ramp Pattern
<img src="https://static.takeuforward.org/wp/uploads/2023/02/image-20.png" width="100">

```python
l=1
r=1
num=65
for i in range(1, 6):
    for j in range(1, 6):
        if j>=l and j<=r:
            print(f"{chr(num)}",end="")
        else:
            print("",end="")
    
    print("")
    num+=1
    r+=1
```

## Alpha-Hill Pattern
<img src="https://static.takeuforward.org/wp/uploads/2022/08/P17.png" width=100>

```python
l=4
r=4
ind = 4
for i in range(1, 5):
    num=65
    for j in range(1, 8):
        if j>=l and j<=r:
            print(f"{chr(num)}",end="")
            if j<ind:
                num+=1
            else:
                num-=1
        else:
            print(" ",end="")
        
    print("")

    l-=1
    r+=1
```
