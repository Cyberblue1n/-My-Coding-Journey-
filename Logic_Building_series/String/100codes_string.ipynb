## Calculate frequency of characters in a string

```python

def calculate_freq(word):
    count_char = {}
    for i in word:
        if i in count_char:
            count_char[i] = count_char.get(i, 0)+1
        else:
            count_char[i] = 1

    return count_char


a = input("Enter a string: ")
print(calculate_freq(a))
```

## Count the sum of numbers in a string

Input string :- “4PREP2INSTAA6”
Output :- 12

```python
def count_sum(mystr):
    list_str = list(mystr)
    sum_val = 0
    for i in list_str:
        if i.isdigit():
            sum_val+=int(i)

    return sum_val
        


if __name__ == "__main__":
    mystr = "4PREP2INSTAA6"
    print(count_sum(mystr))
```

## Replace a sub-string in a string

```python
def substring(mainstr, str1, str2):
    mainstr = mainstr.replace(str1, str2)
    return mainstr

if __name__ == "__main__":
    mainstr = "helloworld"
    replaced = "world"
    str2 = "buddy"

    print(substring(mainstr, replaced, str2))
```
