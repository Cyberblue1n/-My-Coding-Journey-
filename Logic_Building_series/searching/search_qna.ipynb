## Binary Search to find X in sorted array
given array is = {3, 4, 6, 7, 9, 12, 16, 17} and target = 6.

```python
def binary_search(arr, target):
    l=0
    r=len(arr)-1
    while l<=r:
        mid = (l+r)//2
        if arr[mid]==target:
            return mid+1
        elif arr[mid]>target:
            r = mid-1
        else:
            l = mid+1
        
    return "Not Found"

arr = [3,4,6,7,9,12,16,17]
target = 6
print(binary_search(arr, target))
```
